## Bx-Books Csv file Load

```python
books = pd.read_csv("BX-Books.csv", sep=';', encoding='latin-1', on_bad_lines='skip')
```

```python
books.head()
```

```python
books.iloc[10]['Image-URL-L']
```

```python
books.columns
```

```python
books=books[['ISBN','Book-Title','Book-Author','Year-Of-Publication','Publisher','Image-URL-L']]
```

```python
books=books.rename(columns={'Book-Title':'title','Book-Author':'author','Year-Of-Publication':'year','Publisher':'publisher','Image-URL-L':'image_url'})
```
